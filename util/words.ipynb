```python
import json
# load all words fform allWords.json uie the "word" field
# Read, sort, and store
all_words = []


with open('allWords.json', 'r') as f:
    data = json.load(f)
    for entry in data:
        all_words.append(entry['word'].lower())
        all_words = sorted(all_words)
```

```python
# expand umlauts öüä to oe ue ae
expanded_words = []
for word in all_words:
    word = word.replace('ö', 'oe').replace('ä', 'ae').replace('ü', 'ue')
    expanded_words.append(word)

all_words = expanded_words + all_words
# filter to only 5 letter words and remove duplicates
all_words = list(set(all_words))
all_words = [word for word in all_words if len(word) == 5]
```

```python
# Remove all words that contain non abcdefghijklmnopqrstuvwxyz letters
all_words = [word for word in all_words if all(c in 'abcdefghijklmnopqrstuvwxyz' for c in word.lower())]

all_words = sorted(all_words)
```

```python
print(all_words)
len(all_words)
# save to words.json in a structure grouped by first letter
import json
words_dict = {}
for word in all_words:
    first_letter = word[0]
    if first_letter not in words_dict:
        words_dict[first_letter] = []
    words_dict[first_letter].append(word)

with open('wordJson.json', 'w') as f:
    json.dump(words_dict, f, indent=4)
print("Saved to wordJson.json")
```
